```python
#LISTS
numbers=[1,2,3,4,5]
print(numbers)
```

```python
#METHODS IN LISTS
#APPEND
numbers.append(6)
print(numbers)
```

```python
#EXTEND
numbers.extend([7,8,9])
print(numbers)
```

```python
#INSERT
numbers.insert(3,10)
print(numbers)
```

```python
#REMOVE
numbers.remove(10)
print(numbers)
```

```python
#POP
popped_element= numbers.pop(5)
print(popped_element)
```

```python
print(numbers)
```

```python
#INDEX
index= numbers.index(7)
print(index)
```

```python
#COUNT
count= numbers.count(8)
print(count)
```

```python
#SORT
numbers.sort()
print(numbers)
```

```python
#REVERSE
numbers.reverse()
print(numbers)
```

```python
#TUPLES
my_numbers= (1,2,3,4)
print(my_numbers)
```

```python
#METHODS IN TUPLES
#INDEX
index= my_numbers.index(4)
print(index)
```

```python
#COUNT
count= my_numbers.count(1)
print(count)
```

```python
#DICTIONARIES
my_dict= {"name":"Veekee","age":34,"city":"Lagos"}
print(my_dict)
```

```python
#GET
age= my_dict.get("age")
print(age)
```

```python
#GET
name= my_dict.get("name")
print(name)
```

```python
#ITEMS
items= my_dict.items()
print(items)
```

```python
#KEYS
keys= my_dict.keys()
print(keys)
```

```python
#VALUES
values= my_dict.values()
print(values)
```

```python
#UPDATE
my_dict.update({"gender":["Female"]})
print(my_dict)
```

```python
#DICTIONARIES WITH MULTIPLE VALUES
my_dict= dict([('name',['Bambi','Michelle','Raahat']),('age',[21,36,50]),('city',['Budapest','Bern','London'])])
print(my_dict)
```
