```python
#IDENTITY OPERATORS: WITH IDENTITY OPERATORS WE ARE LOOKING AT THE STORAGE OR MEMORY RATHER THAN THE MEMBERS.
#WITH IDENTITY OPERATORS WE WANT TO FIND OUT IF THE TWO OBJECTS ARE THE SAME IN MEMORY,BUT NOT IN VALUE.
# WE HAVE TWO IDENTITY OPERATORS AND THEY ARE:
#1.IS
# 2.IS NOT
```

```python
a=(1,2,3,4,5)
b=(1,2,3,4,5)
```

```python
print(a==b)           # the output of this we give us a TRUE because the values in the objects a and b are the the same
```

```python
print(a is b)       # this produced a FALSE output because the memory of object a and b are not the same regardless
                    # of the value of both objects.
```

```python
print(a is not b)    # the output produced is TRUE because object a is not in the same memory or storage as object b.
```

```python
name= "Dangote INC"
name
```

```python
#METHODS USED IN STRINGS
```

```python
#CONVERTS UPPERCASE: with this you want to convert all the alphabets used in the spelling of "Dangote INC" to uppercase letters.
#upper()
name= name.upper()
name
```

```python
#ISALPHA: THIS CHECKS TO SEE IF THE STRING IS ALL IN ALPHABETS
name= name.isalpha()     # this produces a boolean output of true or false.
                         # the FALSE output produced is because the collection of strings in "Dangote INC" are not all alphabets since we have spaces.
name
```

```python
#CONVERT TO LOWERCASE
#lower()
name= name.lower()
name
```

```python
#ISDIGIT: THIS CHECKS TO SEE IF THERE ARE ANY DIGITS FOUND IN THE STRING "Dangote INC"
#isdgit()
name= name.isdigit()   # this gave a FALSE output because there are no digits found in the string "Dangote INC".
name
```

```python
#ISLOWER: IS CHECKS TO SEE IF ALL THE ALPHABETS IN THE STRING "Dangote INC" is lower case letters.
#islower()
name= name. islower()    # this produced a FALSE output because some of the letters in the string "Dangote INC" are upper case letters.
name
```

```python
#ISUPPER: THIS ALSO CHECKS TO SEE IS THE LETTERS IN THE STRING ARE ALL UPPERCASE LETTERS
#isupper()
name= name.isupper()  # this produced a FALSE output because some not all of the letters used in the string are uppercase letters.
name
```
