```python
# MATPLOTLIB: IT IS A PYTHON LIBRARY WHERE WE CAN PRESENT OUR DATA IN VISUAL FORM.
# WE HAVE TYPES OF MATPLILIB VISUALIZATIONS
#1. QUALITATIVE VISUALIZATION
#A. BAR CHART
# WR ALSO HAVE TO CALL THE OTHER PYTHON LIBRARIES THAT WILL BE NEEDED IN THIS SESSION.
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Loading the Iris dataset
iris = load_iris()
iris
```

```python
# Moving to the visualization aspect:
# we name x to be the iris data which is the independent variable
# and y is the iris.target is the dependent variable
# and our target_names = iris.target_names
```

```python
# Loading the data
x = iris.data
y = iris.target
target_names = iris.target_names


plt.figure(figsize =(8,6))
unique,counts=np.unique(y ,return_counts=True)
plt.bar(target_names,counts,color=plt.cm.viridis(np.linspace(0,1, len(target_names))))
plt.title('Bar Plot of Iris Species')
plt.xlabel('Species')
plt.ylabel('Count')
plt.show()
```

```python
x.shape
```

```python
# PIE CHART:
# THE ONLU THING WE CHANGE HERE IS THE plt.pie
plt.figure(figsize=(8,6))
plt.pie(np.bincount(y), labels= target_names, autopct='%1.1f%%', startangle= 140)
plt.title('Pie plot of Iris Species')
plt.show()
```

```python
# Loading the data
x = iris.data
y = iris.target
target_names = iris.target_names

#Heatmap(Correlation Matrix): The heatmap is going to find the correlation among the species of flowers.
# So hence, we have the correlation matrix:
# We also call our correlation matrix here which we will be using subsequently
# Heatmap(Correlation Matrix)
plt.figure(figsize=(8,6))
corr_matrix= np.corrcoef(x.T)
# So then we plot the heatmap
# Plot heatmap
plt.imshow(corr_matrix, cmap='coolwarm',aspect='auto')
# We are adding annotation to the plotting of the heatmap
# Add Annotations
for i in range(corr_matrix.shape[0]):
  for j in range(corr_matrix.shape[1]):
     plt.text(j, i,f'{corr_matrix[i,j]:.2f}',ha='center', va='center',color='black')
# We add a colorbar
# Add colorbar
plt.colorbar(label='Correlation Coeffeicient')
# We set labels for the heatmap
# Set labels
plt.xticks(ticks=np.arange(len(iris.feature_names)),labels=iris.feature_names)
plt.yticks(ticks=np.arange(len(iris.feature_names)), labels=iris.feature_names)

 # we add a title
 # Add title
plt.title('Heatmap of Correlation Matrix(Iris Dataset)')

# Show Plot
plt.show()

```

```python
# Line Plot
plt.figure(figsize=(8,6))
plt.plot(x[:,0], label=iris.feature_names[0])
plt.xlabel('Sample Index')
plt.ylabel('Sepal length(cm)')
plt.title('Line Plot of Sepal Length(cm)')
plt.legend()
plt.show()
```

```python
# ScatterPlot: useful for showing relationships between continuous variables.
# This time we are plotting the sepal length against the sepal width.
plt.figure(figsize=(8,6))
for i, species in enumerate(target_names):
    plt.scatter(x[y==i, 0], x[y==i, 1], label=species)
plt.xlabel('Sepal Length(cm)')
plt.ylabel('Sepal Width(cm)')
plt.title('Scatter Plot of Sepal Length vs Sepal Width')
plt.legend()
plt.show()
```

```python
# BoxPlot
# Loading the data
x = iris.data
y = iris.target
target_names = iris.target_names

# Boxplot
plt.figure(figsize=(8,6))
colors=plt.cm.viridis(np.linspace(0,1,len(target_names)))  # Generate the colors from the 'viridis' colormap
data_to_plot= [x[y==i,2] for i in range (len(target_names))] # Define data_to_plot before using it
box = plt.boxplot(data_to_plot, patch_artist=True)
for patch, color in zip(box['boxes'],colors): # Corrected typo: b0x to box
  patch.set_facecolor(color)
plt.xlabel('Species')
plt.ylabel('Petal Length(cm)')
plt.title('Box Plot of Petal Length(cm) by Species')
plt.xticks(ticks=np.arange(len(target_names)),labels=target_names)
plt.tight_layout()
plt.show()
```

```python
# Violin Plot: Similar to box plot but this shows the kernel densities that box plot does not show.
# The kernel density shows us where the data is concentrated.

# Plot a Violin Plot
plt.figure(figsize=(8,6))
data_to_plot = [x[y==i,3] for i in range(len(target_names))]
parts = plt.violinplot(data_to_plot, positions=np.arange(len(target_names)), widths=0.6, showmeans=False, showmedians=True)

# Add Color
colors= plt.cm.viridis(np.linspace(0,1,len(data_to_plot)))
for i, pc in enumerate(parts['bodies']):
  pc.set_facecolor(colors[i])
  pc.set_edgecolor('black')
  pc.set_alpha(0.7)

plt.xlabel('Species')
plt.ylabel('Petal Width(cm)')
plt.title('Violin Plot of Petal Width(cm) by Species')
plt.xticks(ticks=np.arange(len(target_names)),labels=target_names)
plt.savefig('violin.png')
plt.show()
```

```python
# Histogram
#Plot a Histogram
plt.figure(figsize=(8,6))
plt.hist(x[:,0], bins=20,alpha=0.6,color='violet', label='Sepal Length')
plt.xlabel('Sepal Length(cm)')
plt.ylabel('Frequency')
plt.title('Histogram of Sepal Length(cm)')
plt.legend()
plt.show()
```
