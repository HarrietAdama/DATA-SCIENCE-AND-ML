```python
# DATA LOADING

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Loading the titanic Dataset
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
titanic = pd.read_csv(url)
titanic
```

```python
# Displaying the first few rows of the dataset
titanic.head()
```

```python
#2. Intial Data Inspection: inspect the basic structure and summary of the dataset.
# This tells us about the data. You can find missing values and other issues wrong with the data from using
# the .info function
# Displaying basic information about the dataset
# we use the .info function
titanic.info()
```

```python
# Displaying summary statistics
titanic.describe(include='all')
```

```python
# SUMMARY STATISTICS: TO GENERATE SUMMARY DATA FOR BOTH CATEGORICAL AND NUMERICAL DATA WE USE:
numerical_summary= titanic.describe()
numerical_summary
```

```python
# CATEGORICAL DATA
categorical_summary = titanic.describe(include=['O'])
categorical_summary
```

```python
#MISSING VALUES ANALYSIS: TO FIND THE MISSING VALUES IN THE DATASET WE DO THIS:
# TO SHOW THE VISUAL REPRESENTATION OF MISSING VALUES, WE CAN USE:
plt.figure(figsize=(10,6))
sns.heatmap(titanic.isnull(), cbar=False, cmap='viridis')
plt.title('Missing Values in Titanice Dataset')
plt.show()



```

```python
#MISSING VALUES ANALYSIS: TO FIND THE MISSING VALUES IN THE DATASET WE DO THIS:
# USE THE .isnull() function to find missing values in the dataset
missing_values = titanic.isnull().sum()
missing_values
```

```python
# DATA CLEANING: Handle missing vlaues by imputing or dropping them as apporpriate.
# If we are dropping some datasets, it should not  be more than 20% of your dataset.
# Filling it with mode or median.
# When it is NUMERICAL DATA it is advisable to use the MEAN OR THE MEDIAN
# When it is CATEGORICAL DATA it is advisable to use the MODE
```

```python
# Fill missing values in 'Age' with the median age
titanic['Age'].fillna(titanic['Age'].median(), inplace=True)

# Fill missing values in 'Embarked' with the most frequent value
titanic['Embarked'].fillna('S', inplace=True)

# Drop the 'Cabin' column due too many missing values
titanic.drop(columns='Cabin', inplace=True, errors='ignore')

#Verify that there are no more missing values
print(titanic.isnull().sum())
```

```python
# DATA VISUALIZATION
# We are creating various plots to visualize the data.
# COUNT PLOT FOR SURVIVAL:
# This plot will show the count of passengers who survived or died.
# Usualluy 0 is for those who DIED and 1 is for those who SURVIVED
```

```python
# Count plot for survival
sns.countplot(x='Survived', data=titanic,hue='Survived')
plt.title('Count of Passengers by Survival')
plt.xlabel('Survived')
plt.ylabel('Count')
plt.show()
```

```python
# Bar plot for Passenger Class(Pclass): This plot will show the distribution of passengers across different classes.
```

```python
# Bar Plot by Class
sns.countplot(x='Pclass', data=titanic,hue='Pclass')
plt.title('Passengers Class Distribution')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.show()
```

```python
 # Age Distribution: A histogram to visualize the age distribution of the passengers
 # Using seaborn
sns.histplot(titanic['Age'].dropna(), bins=20, kde=True)
plt.title('Age Distribution Of Passengers')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
# Survival Rate By Passenger Class: A bar plot to show the survival rate based on the passenger class.

 # Survival Rate by Passenger Class

sns.barplot(x='Pclass', y='Survived', data=titanic, hue='Pclass')
plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.show()
```

```python
# Survival Rate by Sex: A bar plot to show the survival rate by sex.

# Survival Rate By Sex
sns.barplot(x='Sex', y='Survived', data=titanic, hue='Sex')
plt.title('Survival Rate by Sex')
plt.xlabel('Sex')
plt.ylabel('Survival Rate')
plt.show()
```

```python
# Create separate plots for males and females
plt.figure(figsize=(10,5))

#Plot for males
plt.subplot(1, 2, 1)
sns.countplot(x='Survived', data=titanic[titanic['Sex'] == 'male'], hue='Survived')
plt.title('Male Survival Count')
plt.xlabel('Survived')
plt.ylabel('Count')


# Plot  for females


plt.subplot(1, 2, 2)
sns.countplot(x='Survived', data=titanic[titanic['Sex'] == 'female'], hue='Survived')
plt.title('Female Survival Count')
plt.xlabel('Survived')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# Survival Rate by Pclass 1, Survival Rate by Pclass 2, Survival Rate by Pclass 3
plt.figure(figsize=(15,5))

plt.subplot(1, 3, 1)
sns.countplot(x='Survived', data=titanic[titanic['Pclass']== 1], hue='Survived')
plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')


plt.subplot(1, 3, 2)
sns.countplot(x='Survived', data=titanic[titanic['Pclass']== 2], hue='Survived')
plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')


plt.subplot(1, 3, 3)
sns.countplot(x='Survived', data=titanic[titanic['Pclass']== 3], hue='Survived')
plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.tight_layout()
plt.show()



```

```python
# BOX PLOT FOR AGE PASSENGER CLASS: This plot will show the distribution of ages across passenger classes.

# Box plot for age by passenger class
sns.boxplot(x='Pclass', y='Age', data=titanic, hue='Pclass')
plt.title('Age Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Age')
plt.show()
```

```python
# Box Plot for Fare by Passenger Class: This plot will show the distribution of fares by the passenger class.

# BOX PLOT FOR FARE BY PASSENGER CLASS

sns.boxplot(x='Pclass', y='Fare', data=titanic, hue='Pclass')
plt.title('Fare Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.show()
```

```python
# BOX PLOT FOR AGE BY SURVIVAL: This plot will show the distribution of ages for passengers who survived and those who didn't.

# Box plot for age by survival

sns.boxplot(x='Survived', y='Age', data=titanic, hue='Survived')
plt.title('Age Distribution by Survival')
plt.xlabel('Survived')
plt.ylabel('Age')
plt.show()
```

```python
# BOX PLOT FOR FARE BY SURVIVAL: This will show the distribution of fares for passengers who survived or didn't.

# Box plot for fare by survival

sns.boxplot(x='Survived', y='Fare', data=titanic, hue='Survived')
plt.title('Fare Distribution by Survival')
plt.xlabel('Survived')
plt.ylabel('Fare')
plt.show()
```

```python
# CORRELATION OF THE NUMERICAL FEATURES

# Correlation heatmap

plt.figure(figsize=(10,6))
numerical_cols= titanic.select_dtypes(include=['float64','int64'])
corr_matrix = numerical_cols.corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm',vmin=1, vmax=1)
plt.title('Correlation Heatmap of Titanic Features')
plt.show()
```
